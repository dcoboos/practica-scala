```scala
##EJERCICIO 6
####Documentación
Caso ganador(26, 20) (Resultado: 20)

Condición cumplida: Se cumple la segunda condición (else if (bust(handA))).

Explicación: handA (26) supera 21 (da true en bust), mientras que handB (20) no supera 21 (da false). Por tanto, al eliminarse handA por hacer bust, gana directamente handB.

Caso ganador(18, 22) (Resultado: 18)

Condición cumplida: Se cumple la tercera condición (else if (bust(handB))).

Explicación: En este escenario es handB (22) la que supera 21, mientras que handA (18) es una puntuación válida. Al excederse handB, el ganador automático es handA.

Caso ganador(24, 25) (Resultado: 0)

Condición cumplida: Se cumple la primera condición principal (if (bust(handA) && bust(handB))).

Explicación: Ambas manos superan el límite de 21 (24 > 21 y 25 > 21), por lo que ambas se pasan de puntuación (bust doble). La regla indica que en este caso se debe devolver 0.

Caso ganador(17, 19) (Resultado: 19)

Condición cumplida: Se ejecuta el bloque final del else (ninguna supera 21).

Explicación: Ninguna de las dos manos supera 21 (bust(17) y bust(19) devuelven false). El código evalúa cuál de las dos es mayor mediante un if interno (19 > 17), devolviendo 19 por ser la puntuación más alta sin pasarse.

Caso ganador(21, 20) (Resultado: 21)

Condición cumplida: Se ejecuta el bloque final del else (ninguna supera 21).

Explicación: Ambas manos son válidas (ninguna hace bust). Al comparar las puntuaciones, 21 es mayor que 20, por lo que se proclama ganadora a la mano A.
```

```scala
def ganador(handA: Int, handB: Int): Int = {
  if (bust(handA) && bust(handB)) {
    0
  } else if (bust(handA)) {
    handB
  } else if (bust(handB)) {
    handA
  } else {
    if (handA > handB) handA else handB
  }
}
println(ganador(26, 20)) 
println(ganador(18, 22)) 
println(ganador(24, 25)) 
println(ganador(17, 19)) 
println(ganador(21, 20)) 
```

```scala
##EJERCICIO 7
#### Documentación
### Documentación del Ejercicio 7

* **¿Por qué los elementos del array pueden cambiar aunque la variable se haya declarado con `val`?**
  * La palabra reservada `val` garantiza la **inmutabilidad de la referencia**, lo que significa que la variable `jugadores` no puede ser asignada a otro array diferente. Sin embargo, el objeto `Array` en Scala es mutable en su contenido interno; por lo tanto, sus elementos individuales (las posiciones internas del array) sí pueden ser modificados tras su creación.

* **¿Por qué Scala no permite introducir el valor 500 en un `Array[String]`?**
  * Debido al **tipado estático fuerte** de Scala. Al inicializar el array con cadenas de texto (`"Alex"`, `"Chen"`, `"Marta"`), el compilador infiere automáticamente que el tipo de datos es `Array[String]`. El número `500` es de tipo `Int`, por lo que Scala bloquea la asignación al detectar una incompatibilidad de tipos, evitando errores en tiempo de ejecución.

* **Diferencia entre reasignar la variable y modificar un elemento del array:**
  * **Reasignar la variable:** Consiste en cambiar a qué objeto apunta la variable (por ejemplo, intentar hacer `jugadores = Array("Nuevo")`). Esto está prohibido con `val` porque rompería la inmutabilidad de la referencia.
  * **Modificar un elemento:** Consiste en alterar el contenido interno del objeto al que ya apunta la variable (por ejemplo, `jugadores(0) = "Sindhu"`). Esto sí está permitido porque la referencia (`val`) no cambia, solo los datos almacenados dentro de su estructura mutable.
```

```scala
val jugadores = Array("Alex","Chen","Marta")
println(s"Imprimo el array: ${jugadores.mkString(",")}")
jugadores(0)="Sindhu"
println(s"Imprimo el array nuevamente: ${jugadores.mkString(",")}")
jugadores(1)=500
```

```scala
##EJERCICIO 12
#### Documentación
| Expresión | Resultado | Explicación |
| :--- | :---: | :--- |
| `handA > handB` | `false` | Compara si 18 es mayor que 21, lo cual es falso. |
| `handB == 21` | `true` | Comprueba la igualdad exacta entre 21 y 21, devolviendo verdadero. |
| `handC != 21` | `true` | Valida que 25 sea diferente de 21, lo cual se cumple. |
| `(handA <= 21) && (handB <= 21)` | `true` | Ambas manos son menores o iguales a 21 ($18 \le 21$ y $21 \le 21$), por lo que el operador lógico `&&` da verdadero. |
| `(handA > 21) || (handC > 21)` | `true` | Aunque `handA` no supera 21, `handC` (25) sí lo hace; al ser una disyunción (`||`), basta con que una se cumpla. |
| `!(handB > 21)` | `true` | Evalúa si 21 es mayor que 21 (`false`) y aplica el operador de negación `!`, invirtiéndolo a verdadero. |
```

```scala
val handA = 18
val handB = 21
val handC = 25

// 1. ¿Es handA mayor que handB?
val exp1 = handA > handB

// 2. ¿Es handB igual a 21?
val exp2 = handB == 21

// 3. ¿Es handC diferente de 21?
val exp3 = handC != 21

// 4. ¿Son handA y handB menores o iguales a 21?
val exp4 = (handA <= 21) && (handB <= 21)

// 5. ¿Alguna de las manos handA o handC supera 21?
val exp5 = (handA > 21) || (handC > 21)

// 6. ¿No se ha pasado handB de 21? (es decir, no es mayor que 21)
val exp6 = !(handB > 21)

println(s"1. handA > handB: $exp1")
println(s"2. handB == 21: $exp2")
println(s"3. handC != 21: $exp3")
println(s"4. handA y handB <= 21: $exp4")
println(s"5. handA o handC > 21: $exp5")
println(s"6. No se ha pasado handB: $exp6")
```

```scala
##EJERCICIO 13
####Documentación
* **Comparación entre la Versión A (`while`) y la Versión B (`foreach`):**

  * **¿Cuál necesita un contador?**
    * La **Versión A (`while`)** necesita obligatoriamente un contador (por ejemplo, `var i = 0`) para controlar manualmente el índice numérico de la posición actual dentro del array y avanzar paso a paso. La **Versión B (`foreach`)** no requiere contador, ya que la propia abstracción del método gestiona la iteración de forma interna.

  * **¿Cuál necesita una variable `var` para recorrer la colección?**
    * La **Versión A (`while`)** requiere declarar una variable mutable (`var i = 0`) que debe ir incrementándose en cada iteración (`i += 1`). Por el contrario, la **Versión B (`foreach`)** trabaja de forma completamente inmutable y declarativa, sin necesidad de declarar variables de estado externas.

  * **¿Cuál se aproxima más al estilo funcional presentado en el material del curso?**
    * La **Versión B (`foreach`)** se aproxima claramente al estilo funcional. En lugar de indicarle al ordenador *cómo* debe recorrer la estructura paso a paso (enfoque imperativo del `while`), el método `foreach` delega el control de flujo y aplica una función como valor de orden superior directamente sobre cada elemento de la colección.
```

```scala
val manos = Array(17, 24, 21, 26, 18)

// ==========================================
// Versión A: Utilizando un bucle while
// ==========================================
println("--- Versión A (Bucle while) ---")
var i = 0
while (i < manos.length) {
  val mano = manos(i)
  println(s"Mano: $mano -> ¿Es bust?: ${bust(mano)}")
  i += 1
}

// ==========================================
// Versión B: Utilizando foreach con funciones
// ==========================================
println("\n--- Versión B (foreach) ---")
manos.foreach { mano =>
  println(s"Mano: $mano -> ¿Es bust?: ${bust(mano)}")
}

// (Alternativa aún más compacta pasando la función directamente)
// manos.foreach(mano => println(s"Mano: $mano -> ¿Es bust?: ${bust(mano)}"))
```

```scala
##EJERCICIO 14
#### Documentación
* **¿Por qué la función modifica una variable situada fuera de su ámbito local?**
  * La variable `total` está declarada en un ámbito superior (fuera del cuerpo de la función). En Scala, las funciones tienen visibilidad de los elementos de su entorno contenedor (ámbito léxico). Como `total` es una variable mutable (`var`), la función no solo puede leerla, sino también reasignar su valor directamente en memoria cada vez que se ejecuta.

* **Identificación del efecto secundario (*side effect*):**
  * El **efecto secundario** radica en que la función modifica un estado compartido o externo (`total`) en lugar de limitarse a calcular y devolver un resultado basado exclusivamente en sus parámetros de entrada. En la programación puramente funcional se evitan estos comportamientos, ya que dificultan predecir el estado del programa y rastrear errores.
| Característica / Criterio | Función con Estado (`sumarAlTotal`) | Función Pura (`sumar`) |
| :--- | :--- | :--- |
| **Efectos secundarios** | **Sí**. Modifica una variable externa mutable (`var total`). | **No**. No altera ningún estado fuera de su ámbito local. |
| **Transparencia referencial** | **No**. Si ejecutas la función con los mismos parámetros varias veces, el resultado cambia dependiendo del historial de ejecuciones previas. | **Sí**. Ante los mismos parámetros de entrada, siempre devolverá exactamente el mismo resultado. |
| **Facilidad de pruebas y depuración** | **Baja**. Requiere conocer y rastrear el estado global de la aplicación antes de ejecutarla. | **Alta**. Es completamente autocontenida, predecible y muy sencilla de probar de forma aislada. |
```

```scala
var total = 0

def sumarAlTotal(valor: Int) = {
  total = total + valor
}

println(s"Valor inicial de total: $total")

sumarAlTotal(5)
println(s"Después de sumar 5 -> total = $total")

sumarAlTotal(10)
println(s"Después de sumar 10 -> total = $total")

sumarAlTotal(3)
println(s"Después de sumar 3 -> total = $total")
```
